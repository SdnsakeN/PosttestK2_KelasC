```python
import pandas as pd
import seaborn as sns
import numpy as np 
import matplotlib.pyplot as plt

#Bandingkan hubungan antara jumlah kamar tidur (bedrooms) dengan kamar mandi (bathrooms). Tampilkan menggunakan Scatter Plot / Jitter Plot

# ── Load data ──────────────────────────────────────────────────────────────────
df = pd.read_csv("Kelas C_Housing.csv")
 
# ── Plot (pendekatan objek) ────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(8, 6))
 
# Jitter: tambah noise kecil agar titik tidak tumpang tindih
np.random.seed(42)
jitter_x = df["bedrooms"] + np.random.uniform(-0.2, 0.2, len(df))
jitter_y = df["bathrooms"] + np.random.uniform(-0.2, 0.2, len(df))
 
ax.scatter(
    jitter_x,
    jitter_y,
    alpha=0.5,
    edgecolors="white",
    linewidths=0.4,
    s=60,
    c=df["bedrooms"],
    cmap="Blues",
)
 
# Garis tren (regresi linear)
m, b = np.polyfit(df["bedrooms"], df["bathrooms"], 1)
x_line = np.linspace(df["bedrooms"].min(), df["bedrooms"].max(), 100)
ax.plot(x_line, m * x_line + b, color="tomato", linewidth=2, label=f"Tren: y = {m:.2f}x + {b:.2f}")
 
# Label & styling
ax.set_title("Hubungan Jumlah Kamar Tidur vs Kamar Mandi (Jitter Plot)", fontsize=13, fontweight="bold", pad=14)
ax.set_xlabel("Jumlah Kamar Tidur (Bedrooms)", fontsize=11)
ax.set_ylabel("Jumlah Kamar Mandi (Bathrooms)", fontsize=11)
ax.set_xticks(sorted(df["bedrooms"].unique()))
ax.set_yticks(sorted(df["bathrooms"].unique()))
ax.legend(fontsize=10)
sns.despine(ax=ax)
 
plt.tight_layout()
plt.savefig("kategori_B_bedrooms_vs_bathrooms.png", dpi=150, bbox_inches="tight")
plt.show()
```
